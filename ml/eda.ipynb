# Cardio-AI dataset inspection

This notebook profiles the verified data sheet without editing the raw workbook. Target codes and categorical spellings are shown as observed; numeric codes are not assigned undocumented meanings.

```python
import pandas as pd
from ml.constants import TARGETS
from ml.data import load_dataset

df = load_dataset()
print('Shape:', df.shape)
print('Duplicate rows:', df.duplicated().sum())
print('Missing values:', int(df.isna().sum().sum()))
display(df.dtypes.rename('dtype').to_frame())
```

```python
profile = pd.DataFrame({
    'dtype': df.dtypes.astype(str),
    'missing': df.isna().sum(),
    'unique_values': df.nunique(dropna=False),
}).sort_index()
display(profile)
```

```python
for key, spec in TARGETS.items():
    counts = df[spec['column']].value_counts(dropna=False)
    summary = pd.DataFrame({'count': counts, 'percent': counts / len(df) * 100})
    print(f"{key}: {spec['column']} (positive={spec['positive_label']})")
    display(summary)
```
